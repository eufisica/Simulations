# Simulação do Movimento Circular Uniforme (MCU) em Órbitas Planetárias

Este Notebook apresenta uma simulação interativa da mecânica orbital de um satélite artificial em **Movimento Circular Uniforme (MCU)**. O utilizador pode explorar como as propriedades físicas mudam ao orbitar diferentes corpos celestes (Terra, Lua, Marte e Júpiter), alterando a altura da órbita e a massa do satélite.

---

## 1. Fundamentos Teóricos

O movimento de um satélite em torno de um planeta é governado pela **Lei da Gravitação Universal de Newton** e pelas leis do **MCU**. Para que uma órbita seja perfeitamente circular, a força gravítica exercida pelo planeta atua exclusivamente como a **força centrípeta** do sistema.

### 1.1. Raio da Trajetória ($r$)
O raio total da órbita é medido a partir do centro do planeta, sendo a soma do raio do próprio planeta ($R_{planeta}$) com a altitude/altura do satélite ($h$):
$$r = R_{planeta} + h$$

### 1.2. Força Gravítica ($F_g$)
A força de atração entre o planeta de massa $M$ e o satélite de massa $m$ é dada por:
$$Fg = G \cdot \frac{M \cdot m}{r^2}$$
*(Onde $G = 6.6743 \times 10^{-11} \text{ m}^3\text{kg}^{-1}	ext{s}^{-2}$ é a Constante de Gravitação Universal).*

### 1.3. Velocidade Orbital ($v$)
Como a força gravítica é a única responsável por manter o corpo em movimento circular, igualamos a Força Gravítica à Força Centrípeta ($F_c = \frac{m \cdot v^2}{r}$):
$$G \cdot \frac{M \cdot m}{r^2} = \frac{m \cdot v^2}{r}$$

Isolando a velocidade ($v$), obtemos a velocidade necessária para manter a órbita estável:
$$v = \sqrt{\frac{G \cdot M}{r}}$$
*Nota: Reparou que a velocidade orbital **não depende** da massa do satélite ($m$)? Apenas depende da massa do planeta e do raio da órbita.*

### 1.4. Período Orbital ($T$)
O período é o tempo que o satélite demora a completar uma volta inteira (percorrendo a distância do perímetro $2\pi r$). É calculado por:
$$T = \frac{2\pi r}{v}$$

## 2. Instruções de Utilização da Simulação

Na célula seguinte, ao executar o código Python, irá surgir um painel interativo:
1. **Corpo Celeste:** Escolha o planeta ou lua para alterar instantaneamente a massa ($M$) e o raio de base ($R_{planeta}$).
2. **Altura h (km):** Arraste o controlo para definir a distância do satélite à superfície.
3. **Massa Sat. (kg):** Altere a massa do satélite para observar que a **Força Gravítica** altera proporcionalmente, mas a velocidade e o período permanecem idênticos.

### Legenda dos Vetores no Gráfico:
* 🟢 **Vetor Verde:** Velocidade ($v$), sempre tangencial à trajetória.
* 🔴 **Vetor Vermelho:** Força Gravítica ($F_g$), sempre direcionada para o centro (centrípeta).

```python
# Se não tiveres as bibliotecas instaladas, descomenta as duas linhas abaixo e executa uma vez:
# !pip install matplotlib numpy ipywidgets

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, widgets
from IPython.display import display

# Constante de Gravitação Universal
G = 6.6743e-11

# Dados dos corpos celestes (Massa em kg, Raio em metros, cor para o gráfico)
planetas = {
    "Terra": {"M": 5.972e24, "R": 6371000, "color": '#4b9cd3'},
    "Lua": {"M": 7.342e22, "R": 1737400, "color": '#a1a1a1'},
    "Marte": {"M": 6.39e23, "R": 3389500, "color": '#c1440e'},
    "Júpiter": {"M": 1.898e27, "R": 6991100, "color": '#b07f35'}
}

def simular_mcu(planeta, h_km, m_sat):
    # Obter dados do planeta selecionado
    p = planetas[planeta]
    R_p = p["R"]
    h = h_km * 1000  # Converter altura para metros
    r = R_p + h      # Raio da trajetória
    
    # --- CÁLCULOS FÍSICOS (MCU & GRAVITAÇÃO) ---
    v = np.sqrt((G * p["M"]) / r)
    T = (2 * np.pi * r) / v
    Fg = (G * p["M"] * m_sat) / (r**2)
    
    # --- CONFIGURAÇÃO DO GRÁFICO ---
    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_facecolor('black')
    fig.patch.set_facecolor('#1e1e1e')
    
    # Desenhar a Órbita (Linha tracejada)
    theta = np.linspace(0, 2*np.pi, 200)
    ax.plot(r*np.cos(theta)/1e3, r*np.sin(theta)/1e3, 'w--', alpha=0.4, label="Órbita do Satélite")
    
    # Desenhar o Planeta no centro
    planeta_circ = plt.Circle((0, 0), R_p/1e3, color=p["color"], label=f"Corpo Central: {planeta}")
    ax.add_patch(planeta_circ)
    
    # Desenhar o Satélite (Posição simulada no ângulo 0)
    ax.plot(r/1e3, 0, 'ro', markersize=8, label=f"Satélite ({m_sat} kg)")
    
    # Desenhar os Vetores
    ax.quiver(r/1e3, 0, 0, v/1e3, color='#00ff00', scale=v/1e3*3, label="Vetor Velocidade ($v$)")
    ax.quiver(r/1e3, 0, -r/1e3*0.3, 0, color='#ff3333', scale=1, label="Força Gravítica ($F_g$)")
    
    # Ajustes visuais
    lim = (r * 1.4) / 1e3
    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_aspect('equal')
    ax.grid(color='grey', linestyle=':', alpha=0.3)
    
    ax.tick_params(colors='white')
    ax.set_xlabel("Distância X (km)", color='white', fontsize=11)
    ax.set_ylabel("Distância Y (km)", color='white', fontsize=11)
    plt.title(f"Simulação de Movimento Circular Uniforme em {planeta}", color='white', fontsize=14, pad=15)
    
    # --- CAIXA DE TEXTO COM OS RESULTADOS ---
    info_text = (
        f"📊 RESULTADOS DA SIMULAÇÃO:\n"
        f"---------------------------\n"
        f"R_planeta: {R_p/1e3:,.0f} km\n"
        f"Altura (h): {h_km:,.0f} km\n"
        f"Raio (r): {r/1e3:,.0f} km\n\n"
        f"Velocidade (v): {v/1000:.2f} km/s\n"
        f"Período (T): {T/3600:.2f} horas ({(T/60):.1f} min)\n"
        f"Força Grav. (Fg): {Fg:,.2f} N"
    )
    
    ax.text(-lim*1.3, -lim*1.3, info_text, color='white', fontfamily='monospace', fontsize=10,
            bbox=dict(facecolor='#2a2a2a', alpha=0.85, boxstyle='round,pad=0.8', edgecolor='#444444'))
    
    plt.legend(loc='upper right', facecolor='#2a2a2a', edgecolor='#444444', labelcolor='white')
    plt.show()

print("Configura os parâmetros abaixo para atualizar a simulação:")

interact(simular_mcu,
         planeta=widgets.Dropdown(options=list(planetas.keys()), value="Terra", description="Corpo Celeste:"),
         h_km=widgets.IntSlider(min=200, max=150000, step=500, value=5000, description="Altura h (km):", style={'description_width': 'initial'}),
         m_sat=widgets.IntSlider(min=100, max=20000, step=100, value=1500, description="Massa Sat. (kg):", style={'description_width': 'initial'}));
```
