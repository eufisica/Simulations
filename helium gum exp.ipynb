Helium Chewing Gum balloon
Para que um balão de hélio permaneça em equilíbrio no ar (flutue sem subir nem descer), a força de impulsão ($\vec{I}$) deve ser exatamente igual à força gravidade total ($\vec{P}_t$) que atua sobre o sistema.Considerando um balão perfeitamente esférico de raio $r$, a relação matemática estabelece-se através do Princípio de Arquimedes.
1. Dedução MatemáticaA condição de equilíbrio de flutuação é:
$$I = P_t$$
Onde:
$I$ é a impulsão exercida pelo ar exterior: $I = \rho_{\text{ar}} \cdot V \cdot g$;
$P_t$ é o peso total do balão, que inclui a massa do gás hélio ($m_{\text{He}}$) e a massa da estrutura/carga ($m_{\text{estrutura}}$): $P_t = (m_{\text{He}} + m_{\text{estrutura}}) \cdot g$
Como a aceleração da gravidade ($g$) está presente em ambos os lados, podemos simplificar a equação para o equilíbrio de massas:
$$\rho_{\text{ar}} \cdot V = m_{\text{He}} + m_{\text{estrutura}}$$
Sabendo que a massa do hélio depende da sua densidade ($\rho_{\text{He}}$) e do volume do balão ($m_{\text{He}} = \rho_{\text{He}} \cdot V$), substituímos na equação:
$$\rho_{\text{ar}} \cdot V = \rho_{\text{He}} \cdot V + m_{\text{estrutura}}$$
Isolando a massa da estrutura (a massa útil que o balão pode carregar, incluindo o invólucro):
$$m_{\text{estrutura}} = (\rho_{\text{ar}} - \rho_{\text{He}}) \cdot V$$
Como o volume $V$ de uma esfera em função do seu raio $r$ é dado por $V = \frac{4}{3}\pi r^3$, obtemos a função final:
$$m_{\text{estrutura}}(r) = \frac{4}{3}\pi (\rho_{\text{ar}} - \rho_{\text{He}}) \cdot r^3$$
2. Análise da Função
A função que relaciona a massa máxima da estrutura com o raio é uma função cúbica do tipo $m(r) = k \cdot r^3$.
Constante de Proporcionalidade ($k$)
Em condições normais de pressão e temperatura (CNPT: $0^\circ\text{C}$ e $1\text{ atm}$):
Densidade do ar: 
$\rho_{\text{ar}} \approx 1,293\text{ kg/m}^3$
Densidade do hélio: $\rho_{\text{He}} \approx 0,178\text{ kg/m}^3$
$$k = \frac{4}{3}\pi (1,293 - 0,178) \approx 4,67\text{ kg/m}^3$$
Assim, a equação numérica aproximada para o equilíbrio em kg (com $r$ em metros) é:
$$m_{\text{estrutura}} \approx 4,67 \cdot r^3$$
3. Conclusões PráticasDependência Cúbica: Se duplicar o raio do balão ($2r$), a capacidade de carga (massa que ele pode sustentar) aumenta 8 vezes ($2^3$).
Limite Mínimo: O raio tem de ser suficientemente grande para que a massa calculada pela função seja superior à massa do próprio plástico/látex do balão. Se o balão for demasiado pequeno, a massa do invólucro supera a força de sustentação gerada pelo volume de hélio.

```python
import numpy as np
import matplotlib.pyplot as plt
from utils_fisica import calcular_massa_estrutura

# 1. Definir o intervalo de raios (ex: de 0 a 10 metros)
raios = np.linspace(0, 10, 100)

# 2. Calcular as massas usando a função vetorizada
massas = calcular_massa_estrutura(raios)

# 3. Configurar e criar o gráfico
plt.figure(figsize=(8, 5))
plt.plot(raios, massas, label=r'$m \approx 4.67 \cdot r^3$', color='blue', linewidth=2)

# Adicionar detalhes estéticos e informativos
plt.title('Relação Massa vs Raio da Estrutura', fontsize=14)
plt.xlabel('Raio ($r$)', fontsize=12)
plt.ylabel('Massa ($m$)', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

# 4. Exibir o gráfico
plt.tight_layout()
plt.show()
```
