# The Eggnaut Challenge: Physics & Data Analysis

## The Physics of the Egg Drop

### 1. Free Fall (Kinematics)
Before the egg hits the ground, it is in free fall. Assuming air resistance is negligible for this model, the final velocity $v_f$ just before impact depends on the drop height $h$ and acceleration due to gravity $g$:

$$v_f = \sqrt{2gh}$$

### 2. The Impact (Impulse & Momentum)
When the capsule hits the ground, its velocity goes from $v_f$ to $0$. The change in momentum ($\Delta p$) is called **Impulse** ($J$):

$$J = \Delta p = m \cdot v_f$$

According to the impulse-momentum theorem, this is also equal to the average impact force $F_{\text{avg}}$ multiplied by the duration of the impact ($\Delta t$):

$$F_{\text{avg}} \cdot \Delta t = m \cdot v_f \implies F_{\text{avg}} = \frac{m \cdot v_f}{\Delta t}$$

> **Key Takeaway:** To decrease the destructive impact force ($F_{\text{avg}}$), we must design a capsule that extends the impact time ($\Delta t$) using crumple zones or parachutes!

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt

# Define Physics Constants
g = 9.81  # Acceleration due to gravity (m/s^2)

# Define Eggnaut Variables (Modify these based on your actual experiment!)
mass_egg = 0.060       # Mass of a large egg (60 grams converted to kg)
mass_capsule = 0.040   # Mass of the protective capsule (40 grams in kg)
total_mass = mass_egg + mass_capsule # Total mass in kg

drop_height = 9.65      # Drop height in meters

print(True) # To confirm cell execution
```

```python
# 1. Calculate impact velocity
v_impact = np.sqrt(2 * g * drop_height)

# 2. Calculate momentum just before impact
momentum = total_mass * v_impact

print(f"--- EGGNAUT SIMULATION RESULTS ---")
print(f"Total Launch Mass: {total_mass * 1000:.1f} grams")
print(f"Impact Velocity:   {v_impact:.2f} m/s")
print(f"Impact Momentum:   {momentum:.3f} kg·m/s")
```

## Graphing Force vs. Impact Time

Let's look at how the average force changes based on how quickly the egg stops. If the capsule is rigid (like bare concrete), $\Delta t$ is incredibly small, making the force massive. If the capsule has a good cushion, $\Delta t$ increases, and the force drops.

An unprotected egg can generally withstand a maximum force of about **25 Newtons** before cracking. Let's plot the curve!

```python
# Create a range of impact durations (from 0.005 seconds to 0.1 seconds)
impact_times = np.linspace(0.005, 0.10, 500)

# Calculate average force for each time duration
impact_forces = momentum / impact_times

# Maximum force an egg can withstand (Threshold)
egg_threshold = 25.0 

# Plotting
plt.figure(figsize=(10, 6))
plt.plot(impact_times * 1000, impact_forces, label='Impact Force', color='royalblue', linewidth=2)
plt.axhline(y=egg_threshold, color='crimson', linestyle='--', label='Egg Cracking Threshold (25N)')

# Formatting the chart
plt.title('Eggnaut Impact Force vs. Cushioning Time', fontsize=14, fontweight='bold')
plt.xlabel('Impact Duration / Cushioning Time (milliseconds)', fontsize=12)
plt.ylabel('Average Impact Force (Newtons)', fontsize=12)
plt.xlim(5, 100)
plt.ylim(0, 150)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=11)

# Highlight the "Safe Zone"
plt.fill_between(impact_times * 1000, 0, egg_threshold, color='green', alpha=0.15, label='Safe Zone')

plt.show()
```
